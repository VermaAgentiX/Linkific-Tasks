## 2. Create — add a student with a function

Functions let us reuse backend logic. This function creates a new student record and adds it to the list.

```python
def add_student(records, name, age, course, email):
    new_id = max((rec["id"] for rec in records), default=0) + 1

    student = {
        "id": new_id,
        "name": name.strip(),
        "age": age,
        "course": course.strip(),
        "email": email.strip().lower(),
    }

    records.append(student)
    return student


created = add_student(students, "Riya Sharma", 22, "Python Development", "RIYA@example.com")
print("Created record:", created)
print("Total records:", len(students))
```

## 3. Read — list and search students

The backend often returns all records or filters them based on a search query.

```python
def list_students(records):
    return records


def search_students(records, query):
    query = query.strip().lower()
    return [
        rec for rec in records
        if query in rec["name"].lower()
        or query in rec["course"].lower()
        or query in rec["email"].lower()
    ]


print("All students:")
for student in list_students(students):
    print(student["id"], "-", student["name"])

print("\nSearch for 'data':")
print(search_students(students, "data"))
```

## 4. Update — change a student's details

Updating means finding a record by ID and changing selected fields.

```python
def update_student(records, student_id, **changes):
    allowed_fields = {"name", "age", "course", "email"}

    for student in records:
        if student["id"] == student_id:
            for field, value in changes.items():
                if field in allowed_fields:
                    student[field] = value.strip() if isinstance(value, str) else value
            return student

    return None


updated = update_student(students, 2, course="Applied Data Science")
print("Updated record:", updated)
```

## 5. Delete — remove a student

Delete a record by its ID. Return the deleted record if found; otherwise return `None`.

```python
def delete_student(records, student_id):
    for index, student in enumerate(records):
        if student["id"] == student_id:
            return records.pop(index)
    return None


deleted = delete_student(students, 3)
print("Deleted record:", deleted)
print("Remaining student IDs:", [student["id"] for student in students])
```

## 6. Input validation

A backend should validate input before saving it. This simplified example checks a few basic rules; a real application should also validate email format and enforce stricter types.

```python
def validate_student(name, age, course, email):
    errors = []

    if not name.strip():
        errors.append("Name is required.")
    if not isinstance(age, int) or age < 1 or age > 120:
        errors.append("Age must be an integer between 1 and 120.")
    if not course.strip():
        errors.append("Course is required.")
    if "@" not in email or "." not in email.split("@")[-1]:
        errors.append("Email does not look valid.")

    return errors


print(validate_student("Nisha", 21, "Computer Science", "nisha@example.com"))
print(validate_student("", -2, "", "not-an-email"))
```

## 7. Dashboard statistics

Your frontend dashboard can display counts calculated by the backend.

```python
def get_stats(records):
    total = len(records)
    courses = {}

    for student in records:
        course = student["course"]
        courses[course] = courses.get(course, 0) + 1

    return {
        "total_students": total,
        "course_count": len(courses),
        "students_by_course": courses,
    }


print(get_stats(students))
```
