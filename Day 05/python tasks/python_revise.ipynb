## 1. Lists in Python

A list stores an ordered collection. Items are accessed by index, starting at `0`.

````python
nums = [10, 20, 30, 40]

print(nums)
print(nums[0])
print(nums[2])
print(len(nums))
````

## 2. Dictionaries in Python

A dictionary stores key-value pairs. Looking up a value by key is typically **O(1) average time**.

````python
student = {
    "name": "Aarav",
    "age": 21,
    "course": "Computer Science"
}

print(student)
print(student["name"])
print(student.get("age"))
print("course" in student)
````

**Expected output**
```text
{'name': 'Aarav', 'age': 21, 'course': 'Computer Science'}
Aarav
21
True
```

## 4. DSA — Missing Number

**Problem:** Given `n` distinct numbers from the range `0` through `n`, return the missing number.

**Approach:** Expected sum is `n * (n + 1) // 2`. Subtract the actual sum.

````python
def missing_number(nums):
    n = len(nums)
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum


print(missing_number([3, 0, 1]))  # 2
print(missing_number([0, 1]))     # 2
print(missing_number([9, 6, 4, 2, 3, 5, 7, 0, 1]))  # 8
````

**Expected output**
```text
2
2
8
```

- Time complexity: **O(n)**
- Extra space complexity: **O(1)**, excluding the input
- Assumption: distinct integers from `0` to `n`, with exactly one missing.

## 5. DSA — Two Sum (brute-force approach)

Check every pair of numbers until a pair matches the target.

````python
def two_sum_brute(nums, target):
    n = len(nums)

    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                return [i, j]

    return []


print(two_sum_brute([2, 7, 11, 15], 9))  # [0, 1]
print(two_sum_brute([3, 2, 4], 6))       # [1, 2]
print(two_sum_brute([1, 2, 3], 7))       # []
````

**Expected output**
```text
[0, 1]
[1, 2]
[]
```

- Time complexity: **O(n²)**
- Extra space complexity: **O(1)**

## 6. DSA — Two Sum (optimized dictionary approach)

Store each number's index. For each number, check whether its complement (`target - num`) has already been seen.

````python
def two_sum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []


print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
print(two_sum([3, 2, 4], 6))       # [1, 2]
print(two_sum([3, 3], 6))          # [0, 1]
print(two_sum([1, 2, 3], 7))       # []
````

**Expected output**
```text
[0, 1]
[1, 2]
[0, 1]
[]
```

- Time complexity: **O(n) average**
- Extra space complexity: **O(n)**
- The returned values are **indices**, not the numbers themselves.

## 7. Trace the optimized Two Sum algorithm

For `nums = [2, 7, 11, 15]` and `target = 9`:

| Index | Number | Complement | Dictionary before check | Result |
|---:|---:|---:|---|---|
| 0 | 2 | 7 | `{}` | Store `2: 0` |
| 1 | 7 | 2 | `{2: 0}` | Found `2`; return `[0, 1]` |

The dictionary lets us check for the complement without scanning every previous number.
