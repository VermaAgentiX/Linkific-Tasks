## 1. Reverse a list
`nums[::-1]` creates a reversed copy. `nums.reverse()` changes the original list in place and returns `None`.

````python
nums = [10, 20, 30, 40, 50]
rev = nums[::-1]
print("Original:", nums)
print("Reversed copy:", rev)
````

````python
nums = [10, 20, 30, 40, 50]
nums.reverse()
print("Reversed in place:", nums)
````

## 2. Find the maximum number
`max(nums)` is the built-in solution. Check for an empty list if the input may be empty.

````python
nums = [12, 5, 27, 9, 18]
if nums:
    print("Maximum:", max(nums))
else:
    print("The list is empty.")
````

### Manual solution

````python
def find_max(nums):
    if not nums:
        return None
    mx = nums[0]
    for num in nums[1:]:
        if num > mx:
            mx = num
    return mx

print(find_max([12, 5, 27, 9, 18]))  # 27
print(find_max([]))                  # None
````

## 3. Merge two sorted arrays
Use two pointers to append the smaller current value. Both input lists must already be sorted. Time: `O(n + m)`.

````python
def merge_sorted(a, b):
    i = 0
    j = 0
    res = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1
    res.extend(a[i:])
    res.extend(b[j:])
    return res

a = [1, 3, 5, 8]
b = [2, 3, 6, 7]
print(merge_sorted(a, b))  # [1, 2, 3, 3, 5, 6, 7, 8]
````

## 4. Remove duplicates while preserving order
A set tracks values already seen; a result list preserves first-occurrence order. Average time `O(n)`, extra space `O(n)`.

````python
def remove_duplicates(nums):
    seen = set()
    res = []
    for num in nums:
        if num not in seen:
            seen.add(num)
            res.append(num)
    return res

nums = [4, 2, 4, 1, 2, 7, 1]
print(remove_duplicates(nums))  # [4, 2, 1, 7]
````

## 5. Fetch API — Frontend theory
Fetch lets browser JavaScript send HTTP requests to a server and read its response.

### GET example
```javascript
async function loadStudents() {
  const response = await fetch("http://127.0.0.1:8000/students");
  if (!response.ok) throw new Error(`HTTP error: ${response.status}`);
  const students = await response.json();
  console.log(students);
}
loadStudents().catch(console.error);
```

### POST example
```javascript
async function createStudent() {
  const response = await fetch("http://127.0.0.1:8000/students", {
    method: "POST",
    headers: { "Content-Type": "application/json" },
    body: JSON.stringify({ name: "Abhishek", age: 23, course: "Computer Science" })
  });
  if (!response.ok) throw new Error(`HTTP error: ${response.status}`);
  console.log(await response.json());
}
createStudent().catch(console.error);
```

HTTP methods: GET retrieves data; POST creates/submits; PUT replaces; PATCH partially updates; DELETE removes. `fetch()` does not reject just because a server responds with 404, so check `response.ok`. Local browser requests may need CORS configured in FastAPI.
