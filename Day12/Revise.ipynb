## 1. JSON parsing in Python

JSON is a text format often used to exchange data between React and FastAPI. Python's built-in `json` module converts between JSON text and Python values.

- `json.loads(text)` parses JSON text into Python data.
- `json.dumps(value)` converts Python data into JSON text.

### Example

````python
import json

txt = '{"id": 7, "name": "Asha", "course": "AI & ML"}'
rec = json.loads(txt)

print(rec)
print("Student:", rec["name"])

out = json.dumps(rec, indent=2)
print(out)
````

## 4. Client-server communication and CORS

In your app:
1. React is the client.
2. FastAPI is the server.
3. React sends an HTTP request to an API URL.
4. FastAPI returns a response, often containing JSON.
5. React parses the response and updates the UI.

**CORS** means Cross-Origin Resource Sharing. Browsers enforce origin rules. An origin includes the scheme, host, and port. For example, `http://localhost:5173` and `http://127.0.0.1:8000` are different origins. The backend must explicitly allow the frontend origin when browser requests cross origins.

CORS is not authentication; it is a browser mechanism controlling which origins can read cross-origin responses.

### Concept check

````python
frontend_origin = "http://localhost:5173"
backend_origin = "http://127.0.0.1:8000"

print("Frontend:", frontend_origin)
print("Backend: ", backend_origin)
print("These are different origins, so browser CORS configuration may be required.")
````

## 5. Fetch vs Axios

Both Fetch and Axios can send HTTP requests from a React app.

| Feature | Fetch | Axios |
|---|---|---|
| Availability | Built into modern browsers | Must be installed as a dependency |
| HTTP errors | Does not automatically reject for HTTP 4xx/5xx; check `response.ok` | Rejects the promise for non-2xx responses by default |
| JSON | Usually call `response.json()` | JSON response data is commonly available in `response.data` |
| Interceptors | Not built in | Supports request/response interceptors |
| Usage | Lightweight and built in | Convenient features for larger API-client setups |

### Fetch example (JavaScript — for reference)
```javascript
const res = await fetch("http://127.0.0.1:8000/students");
if (!res.ok) {
  throw new Error(`HTTP error: ${res.status}`);
}
const data = await res.json();
```

### Axios example (JavaScript — for reference)
```javascript
import axios from "axios";

const res = await axios.get("http://127.0.0.1:8000/students");
const data = res.data;
```

## 6. DSA — Pair Sum

**Problem:** Given a list of numbers and a target, return the indices of two different values whose sum equals the target. Assume exactly one solution exists.

Example: `nums = [2, 7, 11, 15]`, `target = 9` → indices `[0, 1]`.

A hash map lets us check whether the complement has already appeared. This approach runs in expected **O(n)** time and uses **O(n)** extra space.

````python
def pair_sum(nums, target):
    seen = {}

    for idx, num in enumerate(nums):
        need = target - num
        if need in seen:
            return [seen[need], idx]
        seen[num] = idx

    return None


print(pair_sum([2, 7, 11, 15], 9))  # [0, 1]
print(pair_sum([3, 2, 4], 6))       # [1, 2]
````

## 7. DSA — Merge Intervals

**Problem:** Given intervals, merge any overlapping intervals and return the merged intervals sorted by start.

Example: `[[1, 3], [2, 6], [8, 10], [15, 18]]` becomes `[[1, 6], [8, 10], [15, 18]]`.

Approach:
1. Sort intervals by start.
2. Start the result with the first interval.
3. If the next interval overlaps the last result interval, extend its end.
4. Otherwise, append a new interval.

Time complexity: **O(n log n)** due to sorting.

````python
def merge_intervals(intervals):
    if not intervals:
        return []

    intervals = sorted(intervals, key=lambda item: item[0])
    merged = [intervals[0][:]]

    for start, end in intervals[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])

    return merged


print(merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
print(merge_intervals([[1, 4], [4, 5]]))
````
